# 02. Exploratory Data Analysis (EDA)

Univariate and bivariate statistical analysis to identify distribution shapes, target price skewness, correlation matrices, and the predictive influence of hardware specifications.

```python
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sys.path.append('..')

from src.data.feature_engineering import engineer_features
```

### 1. Ingest Cleaned Data and Synthesize Features

```python
cleaned_path = '../data/processed/cleaned_laptops.csv'
if os.path.exists(cleaned_path):
    df = pd.read_csv(cleaned_path)
    df = engineer_features(df)
    print(f'Engineered dataset dimensions: {df.shape}')
    display(df.head())
else:
    print(f'Cleaned dataset not found at {cleaned_path}')
```

### 2. Target Variable Distribution (Price in EUR)

```python
if 'df' in locals():
    plt.figure(figsize=(10, 5))
    sns.histplot(df['price'], kde=True, bins=35, color='#2563eb')
    plt.title('Histogram and KDE of Laptop Market Prices (EUR)')
    plt.xlabel('Price (EUR)')
    plt.ylabel('Frequency')
    plt.grid(True, alpha=0.3)
    plt.show()
    print(df['price'].describe())
```

### 3. Bivariate Relationship: Price vs. System Memory (RAM)

```python
if 'df' in locals():
    plt.figure(figsize=(10, 5))
    sns.boxplot(x='ram_gb', y='price', data=df, palette='Blues_r')
    plt.title('Laptop Price Distribution across RAM Capacities')
    plt.xlabel('RAM (GB)')
    plt.ylabel('Price (EUR)')
    plt.grid(True, alpha=0.3)
    plt.show()
```

### 4. Influence of Display Pixel Density (PPI) on Valuation

```python
if 'df' in locals():
    plt.figure(figsize=(9, 5))
    sns.scatterplot(x='ppi', y='price', hue='is_touchscreen', data=df, alpha=0.7)
    plt.title('Display PPI versus Market Price')
    plt.xlabel('Pixels Per Inch (PPI)')
    plt.ylabel('Price (EUR)')
    plt.grid(True, alpha=0.3)
    plt.show()
```
